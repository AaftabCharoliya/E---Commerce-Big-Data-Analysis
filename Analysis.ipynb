```python
import pandas as pd

df = pd.read_csv("ecommerce_sales_dataset.csv")

df.head()
```

```python
# Dataset Information
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# ==========================================
# STEP 3 - DATA CLEANING
# ==========================================

# 1. Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

# 2. Check missing values
print("Missing Values:")
print(df.isnull().sum())

# 3. Check invalid dates
print("\nInvalid Dates:")
print(df["Order_Date"].isna().sum())

# 4. Check duplicate Order IDs
print("\nDuplicate Order IDs:")
print(df["Order_ID"].duplicated().sum())

# 5. Check numerical columns for negative values
numeric_columns = [
    "Unit_Price",
    "Quantity",
    "Discount",
    "Revenue",
    "Cost",
    "Profit",
    "Shipping_Cost",
    "Shipping_Days"
]

print("\nNegative Values:")
for column in numeric_columns:
    print(column, "=", (df[column] < 0).sum())

# 6. Display data types after cleaning
print("\nUpdated Data Types:")
print(df.dtypes)

# 7. Display first 5 rows
print("\nCleaned Data Preview:")
print(df.head())
```

```python
# Data Quality Summary

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nMissing Values:", df.isnull().sum().sum())

print("Duplicate Rows:", df.duplicated().sum())

print("Duplicate Order IDs:", df["Order_ID"].duplicated().sum())

print("\nNegative Values:")
print("Unit Price:", (df["Unit_Price"] < 0).sum())
print("Quantity:", (df["Quantity"] < 0).sum())
print("Discount:", (df["Discount"] < 0).sum())
print("Revenue:", (df["Revenue"] < 0).sum())
print("Cost:", (df["Cost"] < 0).sum())
print("Profit:", (df["Profit"] < 0).sum())
print("Shipping Cost:", (df["Shipping_Cost"] < 0).sum())
print("Shipping Days:", (df["Shipping_Days"] < 0).sum())
```

```python
# ==========================================
# STEP 4 - PROFIT & LOSS ANALYSIS
# ==========================================

# Create Profit Status
df["Profit_Status"] = df["Profit"].apply(
    lambda x: "Loss" if x < 0 else "Profit"
)

# Basic Profit/Loss Summary
print("===== PROFIT & LOSS SUMMARY =====")

print("\nTotal Orders:", len(df))

print("Profit Orders:", (df["Profit"] > 0).sum())

print("Loss Orders:", (df["Profit"] < 0).sum())

print("Zero Profit Orders:", (df["Profit"] == 0).sum())

# Loss percentage
loss_percentage = (df["Profit"] < 0).mean() * 100

print(f"\nLoss Order Percentage: {loss_percentage:.2f}%")

# Total Revenue
print(f"\nTotal Revenue: {df['Revenue'].sum():,.2f}")

# Total Cost
print(f"Total Cost: {df['Cost'].sum():,.2f}")

# Total Profit
print(f"Total Profit: {df['Profit'].sum():,.2f}")

# Average Profit
print(f"Average Profit per Order: {df['Profit'].mean():,.2f}")

# Minimum Profit
print(f"Highest Loss: {df['Profit'].min():,.2f}")

# Maximum Profit
print(f"Highest Profit: {df['Profit'].max():,.2f}")
```

```python
# ==========================================
# STEP 5 - CATEGORY & PRODUCT ANALYSIS
# ==========================================

# Category Performance
category_analysis = df.groupby("Category").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_values("Profit", ascending=False)

print("===== CATEGORY PERFORMANCE =====")
print(category_analysis.round(2))


# Product Performance
product_analysis = df.groupby("Product_Name").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_values("Profit", ascending=False)

print("\n===== TOP 10 PRODUCTS BY PROFIT =====")
print(product_analysis.head(10).round(2))


# Bottom 10 Products
print("\n===== BOTTOM 10 PRODUCTS BY PROFIT =====")
print(product_analysis.tail(10).round(2))
```

```python
# ==========================================
# STEP 6 - REGION & COUNTRY ANALYSIS
# ==========================================

# Region Performance
region_analysis = df.groupby("Region").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_values("Profit", ascending=False)

print("===== REGION PERFORMANCE =====")
print(region_analysis.round(2))


# Country Performance
country_analysis = df.groupby("Country").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_values("Profit", ascending=False)

print("\n===== TOP 10 COUNTRIES BY PROFIT =====")
print(country_analysis.head(10).round(2))


# Bottom 10 Countries
print("\n===== BOTTOM 10 COUNTRIES BY PROFIT =====")
print(country_analysis.tail(10).round(2))
```

```python
# ==========================================
# STEP 7 - DISCOUNT VS PROFIT ANALYSIS
# ==========================================

# Create Discount Groups
df["Discount_Group"] = pd.cut(
    df["Discount"],
    bins=[-0.01, 0, 0.10, 0.20, 0.30, 1],
    labels=[
        "No Discount",
        "Low (1-10%)",
        "Medium (11-20%)",
        "High (21-30%)",
        "Very High (30%+)"
    ]
)

# Discount Performance
discount_analysis = df.groupby("Discount_Group", observed=True).agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit=("Profit", "mean"),
    Avg_Discount=("Discount", "mean")
)

print("===== DISCOUNT PERFORMANCE =====")
print(discount_analysis.round(2))


# Profit/Loss by Discount Group
discount_profit_loss = df.groupby(
    "Discount_Group",
    observed=True
).agg(
    Profit_Orders=("Profit", lambda x: (x > 0).sum()),
    Loss_Orders=("Profit", lambda x: (x < 0).sum())
)

print("\n===== PROFIT vs LOSS ORDERS BY DISCOUNT =====")
print(discount_profit_loss)


# Loss Rate
discount_profit_loss["Loss_Rate_%"] = (
    discount_profit_loss["Loss_Orders"] /
    (discount_profit_loss["Profit_Orders"] +
     discount_profit_loss["Loss_Orders"])
) * 100

print("\n===== LOSS RATE BY DISCOUNT =====")
print(discount_profit_loss.round(2))
```

```python
# ==========================================
# STEP 8 - CUSTOMER SEGMENT ANALYSIS
# ==========================================

segment_analysis = df.groupby("Customer_Segment").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum"),
    Avg_Profit=("Profit", "mean"),
    Avg_Revenue=("Revenue", "mean")
).sort_values("Profit", ascending=False)

print("===== CUSTOMER SEGMENT PERFORMANCE =====")
print(segment_analysis.round(2))


# Profit vs Loss Orders
segment_profit_loss = df.groupby("Customer_Segment").agg(
    Profit_Orders=("Profit", lambda x: (x > 0).sum()),
    Loss_Orders=("Profit", lambda x: (x < 0).sum())
)

segment_profit_loss["Loss_Rate_%"] = (
    segment_profit_loss["Loss_Orders"] /
    (segment_profit_loss["Profit_Orders"] +
     segment_profit_loss["Loss_Orders"])
) * 100

print("\n===== CUSTOMER SEGMENT PROFIT vs LOSS =====")
print(segment_profit_loss.round(2))
```

```python
# ==========================================
# STEP 9 - SHIPPING ANALYSIS
# ==========================================

shipping_analysis = df.groupby("Shipping_Method").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit=("Profit", "mean"),
    Avg_Shipping_Days=("Shipping_Days", "mean"),
    Avg_Shipping_Cost=("Shipping_Cost", "mean")
).sort_values("Profit", ascending=False)

print("===== SHIPPING METHOD PERFORMANCE =====")
print(shipping_analysis.round(2))


# Shipping Days Analysis

shipping_days_analysis = df.groupby("Shipping_Days").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit=("Profit", "mean"),
    Avg_Shipping_Cost=("Shipping_Cost", "mean")
).sort_index()

print("\n===== SHIPPING DAYS PERFORMANCE =====")
print(shipping_days_analysis.round(2))
```

```python
# ==========================================
# SHIPPING DAYS - COMPLETE SUMMARY
# ==========================================

shipping_days_summary = df.groupby("Shipping_Days").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit=("Profit", "mean"),
    Avg_Shipping_Cost=("Shipping_Cost", "mean")
).sort_index()

print(shipping_days_summary.round(2).to_string())
```

```python
# ==========================================
# STEP 10 - PAYMENT METHOD & ORDER STATUS
# ==========================================

# Payment Method Analysis
payment_analysis = df.groupby("Payment_Method").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit=("Profit", "mean"),
    Quantity=("Quantity", "sum")
).sort_values("Profit", ascending=False)

print("===== PAYMENT METHOD PERFORMANCE =====")
print(payment_analysis.round(2))


# Payment Method Profit vs Loss
payment_profit_loss = df.groupby("Payment_Method").agg(
    Profit_Orders=("Profit", lambda x: (x > 0).sum()),
    Loss_Orders=("Profit", lambda x: (x < 0).sum())
)

payment_profit_loss["Loss_Rate_%"] = (
    payment_profit_loss["Loss_Orders"] /
    (payment_profit_loss["Profit_Orders"] +
     payment_profit_loss["Loss_Orders"])
) * 100

print("\n===== PAYMENT METHOD PROFIT vs LOSS =====")
print(payment_profit_loss.round(2))


# Order Status Analysis
status_analysis = df.groupby("Order_Status").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit=("Profit", "mean"),
    Quantity=("Quantity", "sum")
).sort_values("Orders", ascending=False)

print("\n===== ORDER STATUS PERFORMANCE =====")
print(status_analysis.round(2))
```

```python
# ==========================================
# STEP 11 - TIME-BASED ANALYSIS
# ==========================================

# Year Analysis
year_analysis = df.groupby("Year").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_index()

print("===== YEARLY PERFORMANCE =====")
print(year_analysis.round(2))


# Quarter Analysis
quarter_analysis = df.groupby("Quarter").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_index()

print("\n===== QUARTERLY PERFORMANCE =====")
print(quarter_analysis.round(2))


# Month Analysis
month_analysis = df.groupby("Month").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_index()

print("\n===== MONTHLY PERFORMANCE =====")
print(month_analysis.round(2))
```

```python
# ==========================================
# STEP 12 - CUSTOMER ANALYSIS
# ==========================================

customer_analysis = df.groupby("Customer_ID").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum"),
    Avg_Order_Value=("Revenue", "mean"),
    Avg_Profit=("Profit", "mean")
)

print("===== TOP 10 CUSTOMERS BY PROFIT =====")
print(
    customer_analysis
    .sort_values("Profit", ascending=False)
    .head(10)
    .round(2)
)

print("\n===== TOP 10 CUSTOMERS BY REVENUE =====")
print(
    customer_analysis
    .sort_values("Revenue", ascending=False)
    .head(10)
    .round(2)
)

print("\n===== TOP 10 CUSTOMERS BY ORDERS =====")
print(
    customer_analysis
    .sort_values("Orders", ascending=False)
    .head(10)
    .round(2)
)
```

```python
# ==========================================
# STEP 13 - CUSTOMER VALUE SEGMENTATION
# ==========================================

# Customer-level revenue thresholds
low_threshold = customer_analysis["Revenue"].quantile(0.33)
high_threshold = customer_analysis["Revenue"].quantile(0.66)

def customer_value(revenue):
    if revenue <= low_threshold:
        return "Low Value"
    elif revenue <= high_threshold:
        return "Medium Value"
    else:
        return "High Value"

customer_analysis["Customer_Value"] = (
    customer_analysis["Revenue"]
    .apply(customer_value)
)

# Segment performance
customer_value_analysis = customer_analysis.groupby(
    "Customer_Value"
).agg(
    Customers=("Revenue", "count"),
    Orders=("Orders", "sum"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Revenue=("Revenue", "mean"),
    Avg_Profit=("Profit", "mean")
).sort_values("Revenue", ascending=False)

print("===== CUSTOMER VALUE SEGMENTATION =====")
print(customer_value_analysis.round(2))
```

```python
# ==========================================
# STEP 14 - PROFIT MARGIN ANALYSIS
# ==========================================

# Category Profit Margin
category_margin = df.groupby("Category").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit_Margin=("Profit_Margin_%", "mean")
).sort_values("Avg_Profit_Margin", ascending=False)

print("===== CATEGORY PROFIT MARGIN =====")
print(category_margin.round(2))


# Region Profit Margin
region_margin = df.groupby("Region").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit_Margin=("Profit_Margin_%", "mean")
).sort_values("Avg_Profit_Margin", ascending=False)

print("\n===== REGION PROFIT MARGIN =====")
print(region_margin.round(2))


# Customer Segment Profit Margin
segment_margin = df.groupby("Customer_Segment").agg(
    Orders=("Order_ID", "count"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Avg_Profit_Margin=("Profit_Margin_%", "mean")
).sort_values("Avg_Profit_Margin", ascending=False)

print("\n===== CUSTOMER SEGMENT PROFIT MARGIN =====")
print(segment_margin.round(2))
```
